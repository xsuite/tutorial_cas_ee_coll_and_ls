# Solution 0 — FCC-ee nominal optics

Load the nominal FCC-ee Z lattice and inspect its tunes, chromaticities, and beta functions at the four interaction points. Complete the missing code.

### Load the lattice

```python
import xtrack as xt

env = xt.load('fccee_z.json')
line = env.fccee_p_ring
line.configure_radiation(model=None)
```

### Tunes and chromaticities

```python
tw = line.twiss4d(delta_chrom=1e-6)
print('Tunes:', tw.qx, tw.qy)
print('Chromaticities:', tw.dqx, tw.dqy)
```

### Beta functions at the interaction points

Inspect `s`, `betx`, and `bety` at `ipa`, `ipd`, `ipg`, and `ipj`.

```python
# s, betx, bety in m
tw.rows[['ipa', 'ipd', 'ipg', 'ipj']].cols['s betx bety']
```

### Full-ring optics

Use `tw.plot()` to display the optics around the full ring.

```python
tw.plot()
```

### One experimental straight

Zoom into the straight around `ipd`, between the dispersion suppressors.

```python
start = 'end_ds_start_straight_ipd'
end = 'end_straight_start_ds_ipd'
tw.rows[start:end].plot()
```
