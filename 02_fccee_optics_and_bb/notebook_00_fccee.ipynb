# FCC-ee optics and beam–beam interaction

Compare two optics for the same FCC-ee Z lattice: the nominal optics and a desqueezed optics with five times larger beta functions at the interaction points. The tunes and corrected linear chromaticities have been restored to the same values.

Then use the nominal optics to study the effect of beam–beam interactions and crab sextupoles on particle survival.

Use the notebooks in this folder:

1. [Optics comparison](notebook_01_optics_comparison.ipynb): inspect the optics and compare corrected and natural chromaticities.
2. [Sensitivity to alignment errors](notebook_02_alignment_sensitivity.ipynb): apply the same random offsets to the final doublets and compare the closed orbits.
3. [Beam–beam interaction and dynamic aperture](notebook_03_beambeam_dynamic_aperture.ipynb): compare particle survival with beam–beam interactions and crab sextupoles switched on and off.
4. [Optional: desqueeze the interaction points](notebook_04_optional_desqueeze.ipynb): reproduce the desqueezed optics with open Twiss and matching.
