### Horizontal orbit RMS

Compute the RMS relative to the design orbit, weighted by distance:

$$x_{\mathrm{RMS}}=\sqrt{\frac{1}{C}\int_0^C x(s)^2\,ds}.$$

Use `np.trapezoid` for the integral, and the same expression with $y$ for the vertical plane.

```python
circumference = line.get_length()
rms_x_nominal = np.sqrt(np.trapezoid(tw_nominal.x**2, tw_nominal.s) / circumference)
rms_x_desqueezed = np.sqrt(np.trapezoid(tw_desqueezed.x**2, tw_desqueezed.s) / circumference)
print('Nominal:', rms_x_nominal)  # m
print('Desqueezed:', rms_x_desqueezed)  # m
```
